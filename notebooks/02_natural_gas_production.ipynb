```python
# by examining the heading of data 1 and data 2, we see the total US production is included in data 1, whereas in data 2, it just has the remainder of states that weren't in data 1 - so data 1 is what we want
# now that we know we want to use data 1, just overwrite "production" - "production_1" and "production_2" were just temporary variables for inspecting the workbook
production = pd.read_excel("../data/NG_PROD_SUM_A_EPG0_FPD_MMCF_M.xls",sheet_name="Data 1",header=2) # with header=2 we're actually saying "Don't treat Excel's first row as the column headers. The actual headers are on the third row."
production.head()
```

```python
# select only the dates and the US production and leave the individual state data behind
production = production[["Date", "U.S. Dry Natural Gas Production (MMcf)"]]
production.head()
```

```python
# rename the production column
production = production.rename(columns={"U.S. Dry Natural Gas Production (MMcf)": "production_mmcf"})
production.head()
```

```python
# our goal now is to turn monthly production in mmcf to average daily production in bcfd
# look at each date and determine how many days are in that date's month so we can divide the monthly prodcution by the days of that month to get average daily prodcution
production["days_in_month"] = production["Date"].dt.days_in_month
production.head()
```

```python
# conversion to billion cubic feet per day
production["production_bcfd"] = (production["production_mmcf"] / 1000 / production["days_in_month"])
production.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(production["Date"], production["production_bcfd"])

plt.title("U.S. Dry Natural Gas Production")
plt.xlabel("Date")
plt.ylabel("Production (Bcf/day)")

plt.show()
```

```python
# identify the biggest monthly dips
production["monthly_change_bcfd"] = production["production_bcfd"].diff() # adding a new column named "monthly_change_bcfd" by using the .diff() method which will calculated the difference between the current month's prodcution and the previous month's production
production.nsmallest(10, "monthly_change_bcfd")[["Date", "production_bcfd", "monthly_change_bcfd"]] # essentially says "Give me the 10 rows with the smallest (i.e. most negative) values in monthly_change_bcfd"
```

```python
# here we're exporting production data to newly created folder under "data" called "processed"
# .to_csv() takes a data frame and generates a csv file under the folder you choose
# index=False tells pandas not to write the DataFrame's row numbers (0, 1, 2, 3...) into the CSV. Those numbers aren't actual market data, so we don't need them
production[["Date", "production_mmcf", "production_bcfd"]].to_csv("../data/processed/us_dry_gas_production_monthly.csv",index=False)
```
